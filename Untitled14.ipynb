```python
import math

# Game tree
tree = [
    [[3, 5], [6, 9]],
    [[1, 2], [0, -1]]
]

def alpha_beta(node, depth, alpha, beta, is_max, path=""):

    if depth == 0:
        print(f"{path}Leaf: {node}")
        return node

    if is_max:
        value = -math.inf
        print(f"{path}MAX")
    else:
        value = math.inf
        print(f"{path}MIN")

    for i, child in enumerate(node):

        branch = "   ├── " if i < len(node)-1 else "   └── "

        result = alpha_beta(
            child,
            depth - 1,
            alpha,
            beta,
            not is_max,
            path + branch
        )

        if is_max:
            value = max(value, result)
            alpha = max(alpha, value)
        else:
            value = min(value, result)
            beta = min(beta, value)

        if alpha >= beta:
            remaining = len(node) - i - 1

            if remaining > 0:
                print(path + "   └── ❌ PRUNED (remaining branches)")
            break

    print(f"{path}   => { 'MAX' if is_max else 'MIN' } value = {value}")

    return value


print("\n========== ALPHA-BETA PRUNING ==========\n")

answer = alpha_beta(
    tree,
    3,
    -math.inf,
    math.inf,
    True
)

print("\n========================================")
print("FINAL OPTIMAL VALUE =", answer)

# # resolution:

print("========================================")
```
