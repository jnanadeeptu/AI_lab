```python

import numpy as np

# Step 1: Define the knapsack problem
weights = np.array([2, 3, 4, 5, 1])
profits = np.array([10, 15, 20, 25, 8])

capacity = 10

N = 10
Max_iter = 10
dim = len(weights)

# Step 2: Fitness function
def fitness(position):

    total_weight = np.sum(position * weights)
    total_profit = np.sum(position * profits)

    if total_weight > capacity:
        return -1000

    return total_profit


# Step 3: Initialize wolves
wolves = np.random.randint(0, 2, (N, dim))

alpha_pos = np.zeros(dim)
beta_pos = np.zeros(dim)
delta_pos = np.zeros(dim)

alpha_score = -float("inf")
beta_score = -float("inf")
delta_score = -float("inf")


# Step 4: Start GWO
for iteration in range(Max_iter):

    # Evaluate fitness
    for i in range(N):

        score = fitness(wolves[i])

        if score > alpha_score:

            delta_score = beta_score
            delta_pos = beta_pos.copy()

            beta_score = alpha_score
            beta_pos = alpha_pos.copy()

            alpha_score = score
            alpha_pos = wolves[i].copy()

        elif score > beta_score:

            delta_score = beta_score
            delta_pos = beta_pos.copy()

            beta_score = score
            beta_pos = wolves[i].copy()

        elif score > delta_score:

            delta_score = score
            delta_pos = wolves[i].copy()

    # Step 5: Calculate a
    a = 2 - iteration * (2 / Max_iter)

    # Step 6: Update wolf positions
    for i in range(N):

        X1 = np.zeros(dim)
        X2 = np.zeros(dim)
        X3 = np.zeros(dim)

        for j in range(dim):

            # Alpha
            A1 = 2 * a * np.random.random() - a
            C1 = 2 * np.random.random()

            D1 = abs(C1 * alpha_pos[j] - wolves[i][j])
            X1[j] = alpha_pos[j] - A1 * D1

            # Beta
            A2 = 2 * a * np.random.random() - a
            C2 = 2 * np.random.random()

            D2 = abs(C2 * beta_pos[j] - wolves[i][j])
            X2[j] = beta_pos[j] - A2 * D2

            # Delta
            A3 = 2 * a * np.random.random() - a
            C3 = 2 * np.random.random()

            D3 = abs(C3 * delta_pos[j] - wolves[i][j])
            X3[j] = delta_pos[j] - A3 * D3

        # Average position
        new_position = (X1 + X2 + X3) / 3

        # Convert continuous values to binary
        probability = 1 / (1 + np.exp(-new_position))

        wolves[i] = (
            np.random.random(dim) < probability
        ).astype(int)

    # Display best profit
    print("Iteration:", iteration + 1,
          "Best Profit:", alpha_score)


# Step 7: Display final result
print("\nBest Selected Items:", alpha_pos)
print("Maximum Profit:", alpha_score)
print("Total Weight:", np.sum(alpha_pos * weights))
```
